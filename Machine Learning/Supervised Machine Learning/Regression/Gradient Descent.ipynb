```python
import numpy as np
```

```python
x = np.array([1,2,3,4,5])
y = 5*x+6
print(x,y)
```

```python
def gradient_descent(x,y,learning_rate=0.01,epochs=3000):
    m=0
    b=0
    for epoch in range(epochs):
        
        y_predicted=m*x+b
        error=y-y_predicted
        mse = np.mean(error**2)
        
        dm = -2 * np.mean(x*error)
        db = -2 * np.mean(error)
        m -= dm * learning_rate
        b -= db * learning_rate
        
        if epoch%100==0 :
            print(f"Cost: {mse}, Slope: {m}, Intercept: {b}")
```

```python
gradient_descent(x,y)
```
